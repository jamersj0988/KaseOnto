```python
from pathlib import Path
import re
import ast

# 設定輸入與輸出路徑。
project_root = Path(r"D:\KaseOnto")
input_path = project_root / "ontology checker" / "default-ontology.js"
output_path = project_root /"kase checker526"/ "ontology_20.dot"

# 讀取 default-ontology.js。
js_text = input_path.read_text(encoding="utf-8")

# 抓出 window.KASEONTO_DEFAULT_ONTOLOGY_DOT = "..."; 的字串內容。
match = re.search(
    r"window\.KASEONTO_DEFAULT_ONTOLOGY_DOT\s*=\s*(?P<dot_string>\".*?\");",
    js_text,
    flags=re.DOTALL,
)

if not match:
    raise ValueError("找不到 window.KASEONTO_DEFAULT_ONTOLOGY_DOT")

# 解開 JS 字串中的 \r\n、\" 等 escape。
dot_text = ast.literal_eval(match.group("dot_string"))

# 統一換行格式，移除多餘空白行，避免輸出莫名其妙的空行。
lines = dot_text.replace("\r\n", "\n").replace("\r", "\n").split("\n")
clean_lines = []
previous_blank = False

for line in lines:
    is_blank = line.strip() == ""

    if is_blank and previous_blank:
        continue

    clean_lines.append(line.rstrip())
    previous_blank = is_blank

# 去掉檔案最前面與最後面的空白行，也不額外補 newline。
clean_dot_text = "\n".join(clean_lines).strip()

# 寫回 DOT 檔案。
output_path.write_text(clean_dot_text, encoding="utf-8")

print(f"已輸出：{output_path}")
```
